```python
from pyspark.sql import SparkSession

# 1. Initialize SparkSession with Delta Lake extensions & packages
spark = (
    SparkSession.builder
    .appName("VerifyBronzeDelta")
    .master("local[*]")
    .config("spark.sql.extensions", "io.delta.sql.DeltaSparkSessionExtension")
    .config("spark.sql.catalog.spark_catalog", "org.apache.spark.sql.delta.catalog.DeltaCatalog")
    .config("spark.jars.packages", "io.delta:delta-spark_2.13:4.4.0")
    .config("spark.driver.host", "127.0.0.1")          # Force local IP binding
    .config("spark.driver.bindAddress", "127.0.0.1")
    .getOrCreate()
)

# 2. Define path to your Bronze Delta table
bronze_path = "/Users/tech/Desktop/Training/nasa-pyspark-pipeline/data/lakehouse/bronze/cmapss_sensors/"

# 3. Read the Delta table
df = spark.read.format("delta").load(bronze_path)

# 4. Inspect schema and record count
print(f"Total Row Count: {df.count():,}")
df.printSchema()

# 5. Preview sample records including ingestion metadata
df.select("unit_id", "time_cycle", "sensor_1", "ingest_timestamp", "source_file").show(5, truncate=False)
```

```python
import os
import glob
from pyspark.sql import SparkSession
import delta

# 1. Locate the Delta Lake JAR installed in your active Python environment
delta_package_path = os.path.dirname(delta.__file__)
jar_pattern = os.path.join(delta_package_path, "lib", "*.jar")
jar_files = glob.glob(jar_pattern)

if not jar_files:
    # Fallback search if structure varies slightly in 4.x
    jar_files = glob.glob(os.path.join(delta_package_path, "**", "*.jar"), recursive=True)

jar_path_str = ",".join(jar_files)
print(f"Loaded local Delta JARs: {jar_files}")

# 2. Build SparkSession with local explicit JAR paths (zero internet dependency)
spark = (
    SparkSession.builder
    .appName("VerifySilverDeltaStandalone")
    .master("local[2]")
    .config("spark.sql.extensions", "io.delta.sql.DeltaSparkSessionExtension")
    .config("spark.sql.catalog.spark_catalog", "org.apache.spark.sql.delta.catalog.DeltaCatalog")
    .config("spark.jars", jar_path_str)
    .config("spark.driver.host", "127.0.0.1")
    .config("spark.driver.bindAddress", "127.0.0.1")
    .getOrCreate()
)

# 3. Read and verify your Silver table
silver_path = "/Users/tech/Desktop/Training/nasa-pyspark-pipeline/data/lakehouse/silver/cmapss_sensors/"
df = spark.read.format("delta").load(silver_path)

print(f"\nSuccessfully Connected! Total Row Count: {df.count():,}")
df.printSchema()
df.show(5, truncate=False)

spark.stop()
```
