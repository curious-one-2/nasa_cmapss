```python
import sys
import os

# 1. Force install delta-spark into the active notebook kernel
print(f"Using Python executable: {sys.executable}")
!{sys.executable} -m pip install delta-spark==4.4.1

# 2. Clean initialization without manual jar hacks
from pyspark.sql import SparkSession
from delta import configure_spark_with_delta_pip

builder = (
    SparkSession.builder
    .appName("VerifySilverDeltaFixed")
    .master("local[2]")
    .config("spark.sql.extensions", "io.delta.sql.DeltaSparkSessionExtension")
    .config("spark.sql.catalog.spark_catalog", "org.apache.spark.sql.delta.catalog.DeltaCatalog")
    .config("spark.driver.host", "127.0.0.1")
    .config("spark.driver.bindAddress", "127.0.0.1")
)

spark = configure_spark_with_delta_pip(builder).getOrCreate()

```

```python
gold_folder = "/Users/tech/Desktop/Training/nasa-pyspark-pipeline/data/lakehouse/gold/fact_engine_telemetry/"

gold_df = spark.read.format("delta").load(gold_folder)

display(gold_df.count)
gold_df.printSchema()
gold_df.show(5, truncate=False)
```

```python
gold_dim_engine_folder = "/Users/tech/Desktop/Training/nasa-pyspark-pipeline/data/lakehouse/gold/dim_engine/"

gold_dim_engine_df = spark.read.format("delta").load(gold_dim_engine_folder)

display(gold_dim_engine_df.count)
gold_dim_engine_df.printSchema()
gold_dim_engine_df.show(5, truncate=False)
```

```python
gold_df.createOrReplaceTempView("fact_engine_telemetry")
gold_dim_engine_df.createOrReplaceTempView("dim_engine")

query = """
    SELECT 
        d.health_status_category,
        COUNT(DISTINCT f.engine_id) AS engine_count,
        ROUND(AVG(d.total_lifespan_cycles), 1) AS avg_lifespan,
        ROUND(AVG(f.sensor_4), 2) AS avg_sensor_4_reading,
        ROUND(AVG(f.sensor_4_roll_mean_5), 2) AS avg_sensor_4_rolling_mean,
        ROUND(AVG(f.rul), 1) AS avg_rul
    FROM fact_engine_telemetry f
    JOIN dim_engine d ON f.engine_id = d.engine_id
    GROUP BY d.health_status_category
"""

spark.sql(query.strip()).show(truncate=False)
```
